## 1. What you will be able to do

Create typed, least-privilege tools with clear semantics for validation, errors, retries, idempotency, side effects, and authorization.

**Phase:** II. Tools and knowledge
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Write Tool Contracts Models Can Use | Contract tests and selection evaluation. |
| 2. Normalize Errors, Timeouts, and Cancellation | Error-mapping test matrix. |
| 3. Make Retries and Side Effects Safe | Proof that repeated delivery causes one intended effect. |
| 4. Enforce Least Privilege and Approval | Authorization table and denied-action traces. |

### 1. Write Tool Contracts Models Can Use

Design names, descriptions, schemas, examples, preconditions, and postconditions that separate model-facing affordances from implementation details. Measure ambiguous tool selection and distinguish discovering a tool from permission to execute it.

### 2. Normalize Errors, Timeouts, and Cancellation

Create a tool-result envelope distinguishing invalid input, denied action, transient failure, permanent failure, partial result, timeout, and cancellation. Prevent raw stack traces or secrets from entering model context.

### 3. Make Retries and Side Effects Safe

Separate queries from commands; introduce idempotency keys, deduplication, retry policies, and compensation. Diagnose a duplicate-charge scenario caused by naive retry.

### 4. Enforce Least Privilege and Approval

Scope credentials and tools by task, resource, action, and time. Add policy checks and human confirmation for high-impact, irreversible, or externally visible operations.

## 4. Challenge preview: Harden a Ticketing Toolset

Repair tools that expose secrets, retry unsafe actions, mix errors with data, and grant excessive access. Pass contract, fault-injection, and authorization tests.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
