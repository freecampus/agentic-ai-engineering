## 1. What you will be able to do

Compare API, DOM, accessibility-tree, and visual control; ground multimodal observations; and constrain dynamic user-interface actions.

**Phase:** V. Advanced capabilities
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Prefer Stable Interfaces Before Pixels | Interface-selection decision and tests. |
| 2. Ground Visual Observations and Actions | Grounding accuracy evaluation. |
| 3. Handle Dynamic Pages, Authentication, and Downloads | Browser-state and download-validation tests. |
| 4. Defend Against Visual Injection and Unsafe UI Actions | Adversarial browser benchmark. |

### 1. Prefer Stable Interfaces Before Pixels

Choose API, structured data, DOM, accessibility tree, or screenshot interaction based on availability and reliability. Measure brittleness under page changes.

### 2. Ground Visual Observations and Actions

Combine screenshots, OCR, element descriptions, coordinates, and action history. Detect stale screenshots and ambiguous targets before clicking.

### 3. Handle Dynamic Pages, Authentication, and Downloads

Wait on observable conditions, preserve session boundaries, validate downloads, and avoid leaking authenticated content. Separate navigation from consequential action.

### 4. Defend Against Visual Injection and Unsafe UI Actions

Test malicious page instructions, hidden text, deceptive controls, and irreversible submissions. Add domain allowlists, action previews, and confirmation.

## 4. Challenge preview: Complete a Sandboxed Web Task Safely

Automate a multi-step task in a local benchmark website using the most stable interfaces available. Hidden variants change layout, inject instructions, delay elements, and present a consequential final action.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
