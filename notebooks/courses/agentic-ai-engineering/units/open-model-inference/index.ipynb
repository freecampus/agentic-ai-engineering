## 1. What you will be able to do

Select, serve, quantize, batch, route, and cache open models using measured quality, license, memory, throughput, latency, privacy, and energy tradeoffs.

**Phase:** VI. Production
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the

Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Match Models and Licenses to Tasks | Model scorecard. |
| 2. Run Quantized Models Locally | Local inference benchmark. |
| 3. Serve Models for Throughput | Load-test curves and saturation point. |
| 4. Route, Cache, and Fall Back Deliberately | Routing policy and failure drills. |

### 1. Match Models and Licenses to Tasks

Build a shortlist from model cards, supported modalities, context, structured output, tool use, language coverage, licenses, and hardware. Verify claims on the course eval suite.

### 2. Run Quantized Models Locally

Use Ollama or llama.cpp, inspect quantization and memory use, and compare CPU/GPU execution. Identify quality losses on the actual task rather than generic prose.

### 3. Serve Models for Throughput

Use a maintained open serving engine such as vLLM or SGLang; study continuous batching, KV cache, parallelism, and backpressure. Treat maintenance-only TGI as a legacy/migration comparison, not the default new deployment.

### 4. Route, Cache, and Fall Back Deliberately

Choose models by task complexity, risk, availability, latency, and privacy. Cache only safe, context-equivalent requests and test fallback semantic differences.

## 4. Challenge preview: Meet a Quality–Latency–Memory Budget

Configure an open-model stack that meets declared thresholds on constrained hardware. Compare at least three model/configuration options and defend the Pareto choice.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
