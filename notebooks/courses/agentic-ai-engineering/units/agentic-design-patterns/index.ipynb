## 1. What you will be able to do

Implement and compare chaining, routing, parallelization, orchestrator–worker, evaluator–optimizer, reflection, and state-machine patterns.

**Phase:** III. Architecture
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Complete the preceding unit, [Build Memory That Earns Its Cost](https://freecampus.github.io/agentic-ai-engineering/courses/agentic-ai-engineering/units/agent-memory/index.html), when its lessons become available. This is a cumulative engineering sequence, not a set of independent tutorials.

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the
shared package. Work in a clean Colab/Jupyter notebook or a local Conda
environment with Poetry-managed dependencies. The core route will not require
provider keys or paid inference.
Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Chain and Route Predictably | Routing matrix and chain failure trace. |
| 2. Parallelize Independent Work | Latency/cost experiment. |
| 3. Orchestrate Workers and Evaluators | Stopping and disagreement tests. |
| 4. Encode Control in State Machines and Graphs | Executable state graph and invariant checks. |

### 1. Chain and Route Predictably

Build typed prompt chains and routers with explicit fallbacks. Measure error propagation and routing confusion against a single-call baseline.

### 2. Parallelize Independent Work

Use concurrency for independent tasks, then merge results with conflict and timeout policies. Distinguish wall-clock improvement from increased total compute.

### 3. Orchestrate Workers and Evaluators

Implement orchestrator–worker and evaluator–optimizer loops with bounded retries. Prevent a model evaluator from becoming an unquestioned source of truth.

### 4. Encode Control in State Machines and Graphs

Represent branches, cycles, interrupts, and terminal states explicitly. Compare code-controlled and model-controlled transitions.

## 4. Challenge preview: Implement One Task Four Ways

Build deterministic, chained, graph, and agent versions of one task. Recommend one using held-out quality, latency, cost, observability, and security evidence.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
