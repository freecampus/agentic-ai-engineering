## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Define Metrics from the Task Contract | Metric specification with gaming risks. |
| 2. Build Representative Evaluation Data | Dataset card and coverage matrix. |
| 3. Combine Deterministic, Model, and Human Graders | Grader agreement and disagreement analysis. |
| 4. Compare Systems with Uncertainty | Reproducible paired comparison with repeated-success metrics, uncertainty, and model/harness ablations. |

### 1. Define Metrics from the Task Contract

Translate success and harm into observable measures. Separate final-answer quality from trajectory correctness, policy compliance, and resource use; verify actual environment state rather than trusting an agent claim that an action completed.

### 2. Build Representative Evaluation Data

Create normal, boundary, adversarial, temporal, and abstention cases; prevent leakage; version fixtures and splits. Analyze whose tasks and failures are missing.

### 3. Combine Deterministic, Model, and Human Graders

Prefer exact and executable checks where possible; design rubrics and blinded pairwise judgments where needed. Calibrate model judges against human labels.

### 4. Compare Systems with Uncertainty

Run repeated trials, estimate confidence intervals, analyze paired differences, and avoid declaring tiny noisy gains. Distinguish pass@k from pass^k, reset trial state, and separate model changes from harness changes when constructing quality-latency-cost frontiers.

## 4. Challenge preview: Build an Evaluation Harness Before Improving the Agent

Given two plausible agents, design data and graders that reveal their real tradeoffs. Hidden cases test leakage, judge bias, variance, and misleading aggregate scores.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
