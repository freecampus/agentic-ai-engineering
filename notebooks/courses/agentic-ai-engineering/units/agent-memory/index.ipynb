## 1. What you will be able to do

Design working, episodic, semantic, and procedural memory, including versioned skills, with controlled writes, retrieval, consolidation, privacy, deletion, and poisoning defenses.

**Phase:** II. Tools and knowledge
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Separate Memory Types and Lifecycles | Memory architecture and retention table. |
| 2. Decide What May Be Written | Memory-write policy tests. |
| 3. Retrieve, Consolidate, and Resolve Contradictions | Longitudinal memory benchmark. |
| 4. Protect, Inspect, Correct, and Delete Memory | Deletion and poisoning audit. |

### 1. Separate Memory Types and Lifecycles

Map working state, event history, user facts, learned procedures, and cached artifacts to distinct stores and lifetimes. Represent reusable procedures as versioned skills, separate them from factual memory and model-weight updates, and avoid treating the entire transcript as memory.

### 2. Decide What May Be Written

Implement consent, validation, salience, confidence, source, TTL, and sensitivity policies. Reject speculative or sensitive facts and stage generated skill changes for review before they can affect future tasks.

### 3. Retrieve, Consolidate, and Resolve Contradictions

Rank memories by relevance, authority, recency, and confidence; consolidate duplicates; surface conflicts. Compare synchronous and background consolidation.

### 4. Protect, Inspect, Correct, and Delete Memory

Add user inspection, correction, export, deletion, and poisoning detection. Verify deletion across indexes, caches, and summaries.

## 4. Challenge preview: Ship a Grounded Knowledge Worker

Extend the research assistant with evaluated hybrid retrieval, bounded memory, provenance, and user-controlled deletion. Show that memory improves a longitudinal task without harming privacy or contradiction handling.

The complete challenge contract, starting fixtures, progressive checks, three-level

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
