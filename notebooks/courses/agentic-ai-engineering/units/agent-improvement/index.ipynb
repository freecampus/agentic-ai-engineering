## 1. What you will be able to do

Optimize prompts, procedures, policies, and small models with train/validation/test separation while detecting overfitting, leakage, reward hacking, and unsafe self-modification.

**Phase:** V. Advanced capabilities
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Complete the preceding unit, [Make Long-Running Agents Durable](https://freecampus.github.io/agentic-ai-engineering/courses/agentic-ai-engineering/units/durable-execution/index.html), when its lessons become available. This is a cumulative engineering sequence, not a set of independent tutorials.

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the
shared package. Work in a clean Colab/Jupyter notebook or a local Conda
environment with Poetry-managed dependencies. The core route will not require
provider keys or paid inference.
Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Optimize Prompts and Programs on Training Data | Optimization log with skill lineage, frozen holdout, ablations, and measured regressions. |
| 2. Learn Routing and Selection Policies | Offline policy evaluation. |
| 3. Fine-Tune Structured Tool Use | Data card and before/after evaluation. |
| 4. Detect Reward Hacking and Unsafe Self-Improvement | Reward-hacking case analysis and controls. |

### 1. Optimize Prompts and Programs on Training Data

Define a search space, objective, constraints, and held-out split. Compare manual and automated prompt/program optimization, including a Hermes-inspired experiment with no skills, a fixed reviewed skill, and a generated skill; distinguish procedural adaptation from weight training.

### 2. Learn Routing and Selection Policies

Use contextual bandit intuition to choose models, tools, or strategies from feedback. Separate exploration from high-risk production actions.

### 3. Fine-Tune Structured Tool Use

Prepare licensed trajectories, format supervised examples, fine-tune a small model where resources permit, and evaluate schema validity and tool choice rather than imitation alone.

### 4. Detect Reward Hacking and Unsafe Self-Improvement

Study process/outcome rewards, credit assignment, evaluator exploitation, test tampering, and optimization pressure. Stage memory and skill changes in isolated candidates; require immutable external gates, independent approval, revocation, and rollback before promotion.

## 4. Challenge preview: Build a Durable Specialist

Create a coding, browser, or data specialist that resumes after failure and evaluates a bounded improvement using a training split. Report held-out gains, regressions, or an inconclusive result without weakening safety gates; promote only evidence-backed candidates.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
