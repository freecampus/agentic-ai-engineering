## 1. What you will be able to do

Package agents behind clear APIs and workers with persistent state, queues, concurrency control, idempotency, streaming progress, and tenant isolation.

**Phase:** VI. Production
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Define Service and Task APIs | OpenAPI contract and consumer tests. |
| 2. Separate API, Workflow, Worker, and Storage Roles | Deployment and data-flow diagram. |
| 3. Control Concurrency, Backpressure, and Failure Cascades | Overload and recovery test. |
| 4. Isolate Tenants, Tasks, and Resources | Isolation test suite. |

### 1. Define Service and Task APIs

Design submit, inspect, stream, cancel, approve, and retrieve-artifact endpoints. Use stable task IDs, idempotency keys, and explicit terminal status.

### 2. Separate API, Workflow, Worker, and Storage Roles

Build a local service topology with API, queue/workflow engine, workers, database, artifact store, and model/tool adapters. Trace ownership of every state transition.

### 3. Control Concurrency, Backpressure, and Failure Cascades

Apply rate limits, queues, worker pools, circuit breakers, bulkheads, and admission control. Diagnose retry storms and overloaded shared dependencies.

### 4. Isolate Tenants, Tasks, and Resources

Propagate identity and quotas, partition data and artifacts, restrict worker capabilities, and test cross-tenant access attempts. Contrast a trusted single-operator/team assistant gateway with an adversarial multi-tenant service; do not infer isolation from separate sessions.

## 4. Challenge preview: Operate a Resumable Multi-User Service

Deploy a local containerized service, submit concurrent tasks, stream progress, cancel one task, restart a worker, and prove state recovery and tenant isolation.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
