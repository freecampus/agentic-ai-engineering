## 1. What you will be able to do

Build and inspect FreeCampusAgent by connecting a language model, conversation memory, tools, and a simple planner; explain how each addition changes what the agent can do.

**Phase:** I. Foundations
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates await a learner pilot.

**Scenario:** Build FreeCampusAgent gradually. Begin with a small class
scaffold, connect a language model, then add conversation memory, a tool, and a
simple planner. In each lesson, inspect the code and the behavior it produces.

## 2. Before you begin

Complete the preceding unit, [Launch a Reproducible Agent Engineering Lab](https://freecampus.github.io/agentic-ai-engineering/courses/agentic-ai-engineering/units/launch-agent-lab/index.html), when its lessons become available. This is a cumulative engineering sequence, not a set of independent tutorials.

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the
shared package. Work in a clean Colab/Jupyter notebook or a local Conda
environment with Poetry-managed dependencies. Lesson 1 is a Python-only scaffold
and needs no API key. Later lessons will state clearly when an external model is
needed. Keep credentials and personal data out of notebooks and submitted
artifacts.

Bring your Unit 0 launch card and evidence journal. You should be able to run a
Python cell, inspect a function, predict its output, and
distinguish a permitted action from a requested one. No new agent framework,
provider SDK, external tool integration, or reinforcement-learning background is
needed. Keep one notebook per lesson and a folder for the final decision record.

**Vocabulary:** agent, language model, message, conversation memory, tool, and
planner. Each term is explained when it first appears. **Risk preview:** the
model can still give a wrong answer, and each added message increases the amount
of text sent in later requests. The message list stays in the running notebook;
this unit does not save it after the session ends.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. What Is an Agent? | FreeCampusAgent's class definition, first output, and explanation of its placeholder methods. |
| 2. Connect an LLM to FreeCampusAgent | One inspected model response and a model-backed FreeCampusAgent answer. |
| 3. Give FreeCampusAgent Conversation Memory | Message history before and after two turns, plus a model-backed follow-up. |
| 4. Give FreeCampusAgent a Tool | A tool call, its Python result, and the reply that uses that result. |
| 5. Add a Planner to FreeCampusAgent | A short plan, step-by-step execution record, and explanation of the limits. |

### 1. What Is an Agent?

Introduce FreeCampusAgent's class scaffold, explain its methods, and run its first task without a connected model or tools.

### 2. Connect an LLM to FreeCampusAgent

Explain how an LLM receives a message and generates a response. Inspect the
returned model ID and token-use fields, then connect one model call to
FreeCampusAgent. This first version makes one call and does not use memory,
tools, or planning.

### 3. Give FreeCampusAgent Conversation Memory

Explain that each model request receives only the messages sent with it. Add a small conversation-memory object to FreeCampusAgent and inspect a follow-up request that includes earlier messages.

### 4. Give FreeCampusAgent a Tool

Connect one read-only course lookup to FreeCampusAgent. Follow the model's structured request, the Python function's result, and the final answer.

### 5. Add a Planner to FreeCampusAgent

Add a simple planner that turns a task into ordered steps, then inspect how FreeCampusAgent handles those steps.

## 4. Challenge preview: De-Agentify an Overbuilt System

Simplify a fictional five-agent customer-request system while preserving required behavior. The solution must demonstrate fewer model decisions, lower expected cost, and equal or better test performance.

The review draft supplies a behavior contract, a five-controller baseline, an
incomplete starter, ordinary/boundary/adversarial fixtures, progressive checks,
exactly three hints, and a hidden worked solution. It requires current public
policy evidence, no forbidden effects, explicit escalation, fewer decisions, and
lower cost under stated synthetic assumptions. It does not require a live model.
The challenge is not released or completable yet. A successful-looking output
alone will not establish reliability: preserve the inputs, limits, observed
behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, cost assumptions, and safety;
identify latency as unmeasured rather than inventing a performance claim.

1. After Lesson 1: stop with the FreeCampusAgent class, its first output, and an explanation of what is still a placeholder.
2. After Lesson 2: stop with an inspected response and one model-backed FreeCampusAgent answer.
3. After Lesson 3: stop with the message list from a two-turn conversation and an explanation of what is sent to the model.
4. After Lesson 4: stop with a tool call and its returned result.
5. After Lesson 5: stop with a short plan and an explanation of how the agent processes its steps.
6. After the challenge: retain the original failures, simpler implementation,
   qualification disclosure, synthetic cost comparison, and one revisit trigger.
