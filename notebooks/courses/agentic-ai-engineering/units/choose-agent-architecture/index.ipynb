## 1. What you will be able to do

Distinguish programs, workflows, agents, and multi-agent systems; define degrees of autonomy; and select the simplest adequate architecture from evidence.

**Phase:** I. Foundations
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates await a learner pilot.

**Scenario:** Lantern repair café has been offered a five-agent customer help
desk. You will inspect who controls its actions, trace what it observes, enforce
a task contract, and replace unnecessary autonomy with a simpler adequate design.
The input is deliberately structured: a supplied topic and an untrusted note.
This unit tests control and evidence, not natural-language understanding.

## 2. Before you begin

Complete the preceding unit, [Launch a Reproducible Agent Engineering Lab](https://freecampus.github.io/agentic-ai-engineering/courses/agentic-ai-engineering/units/launch-agent-lab/index.html), when its lessons become available. This is a cumulative engineering sequence, not a set of independent tutorials.

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the
shared package. Work in a clean Colab/Jupyter notebook or a local Conda
environment with Poetry-managed dependencies. The core route will not require
provider keys or paid inference.
Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

Bring your Unit 0 launch card and evidence journal. You should be able to import
the package, inspect a function, predict a short trace, write an assertion, and
distinguish a permitted action from a requested one. No new agent framework,
provider SDK, external tool integration, or reinforcement-learning background is
needed. Keep one notebook per lesson and a folder for the final decision record.

**Vocabulary:** control authority, workflow, policy, observation, environment
state, terminal outcome, task contract, approval, architecture decision record.
Each term is explained at its point of use. **Risk preview:** a correct-looking
answer can conceal a stale source, a missing observation, an unauthorized request,
or unnecessary work. These offline fixtures enforce narrow dispatch boundaries;
they are not security sandboxes or evidence of live-model reliability.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Separate Automation, Workflows, and Agents | Classification table with disputed cases and defended criteria. |
| 2. Model the Agent–Environment Loop | State-transition trace and diagram. |
| 3. Bound Autonomy with Task Contracts | Validated task-contract schema and examples. |
| 4. Choose the Simplest Adequate Architecture | Architecture decision record with rejection reasons. |

### 1. Separate Automation, Workflows, and Agents

Compare deterministic scripts, DAGs, state machines, LLM workflows, and feedback-driven agents. Classify systems by who chooses the next action, what feedback is available, and how termination is controlled.

### 2. Model the Agent–Environment Loop

Represent observations, internal state, actions, environment transitions, objectives, costs, and terminal states. Trace a small grid-world and a tool-using assistant through the same conceptual loop.

### 3. Bound Autonomy with Task Contracts

Translate user intent into measurable success, permissions, budgets, deadlines, approval gates, escalation, and forbidden outcomes. Identify underspecified contracts that invite unsafe or wasteful behavior.

### 4. Choose the Simplest Adequate Architecture

Compare a rule, retrieval query, workflow, single agent, and multi-agent system for the same scenario. Calculate the additional failure surface introduced by each level of autonomy.

## 4. Challenge preview: De-Agentify an Overbuilt System

Simplify a fictional five-agent customer-request system while preserving required behavior. The solution must demonstrate fewer model decisions, lower expected cost, and equal or better test performance.

The review draft supplies a behavior contract, a five-controller baseline, an
incomplete starter, ordinary/boundary/adversarial fixtures, progressive checks,
exactly three hints, and a hidden worked solution. It requires current public
policy evidence, no forbidden effects, explicit escalation, fewer decisions, and
lower cost under stated synthetic assumptions. It does not require a live model.
The challenge is not released or completable yet. A successful-looking output
alone will not establish reliability: preserve the inputs, limits, observed
behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, cost assumptions, and safety;
identify latency as unmeasured rather than inventing a performance claim.

1. After Lesson 1.1: stop with a classification table and a tested zero-decision replacement.
2. After Lesson 1.2: stop with two checked traces and a missing-observation explanation.
3. After Lesson 1.3: stop with a validated contract and allow/deny/confirm/stop evidence.
4. After Lesson 1.4: stop with per-case comparisons and an architecture decision record, including no-go criteria.
5. After the challenge: retain the original failures, simpler implementation,
   qualification disclosure, synthetic cost comparison, and one revisit trigger.
