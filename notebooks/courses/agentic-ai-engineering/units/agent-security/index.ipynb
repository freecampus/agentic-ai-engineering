## 1. What you will be able to do

Threat-model an agent and defend against injection, poisoning, exfiltration, confused-deputy behavior, command/URL attacks, excessive agency, and supply-chain compromise.

**Phase:** IV. Evaluation and safety
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Complete the preceding unit, [Observe and Debug Nondeterministic Systems](https://freecampus.github.io/agentic-ai-engineering/courses/agentic-ai-engineering/units/observability-debugging/index.html), when its lessons become available. This is a cumulative engineering sequence, not a set of independent tutorials.

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the
shared package. Work in a clean Colab/Jupyter notebook or a local Conda
environment with Poetry-managed dependencies. The core route will not require
provider keys or paid inference.
Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Draw Assets, Actors, Trust Boundaries, and Data Flows | Data-flow diagram and risk register. |
| 2. Defend Against Prompt Injection and Data Exfiltration | Adversarial document suite and containment results. |
| 3. Sandbox Code, Browsing, and Network Access | Sandbox escape test results. |
| 4. Protect Identity, Secrets, Tenants, and Dependencies | Access matrix and supply-chain audit. |

### 1. Draw Assets, Actors, Trust Boundaries, and Data Flows

Create a threat model for user input, retrieved content, model output, tools, memory, protocols, credentials, telemetry, and humans. Prioritize threats by impact and exposure.

### 2. Defend Against Prompt Injection and Data Exfiltration

Treat external text as data, constrain tool capabilities, enforce policy outside the model, label provenance, and test direct and indirect injection. Explain why a stronger system prompt is insufficient.

### 3. Sandbox Code, Browsing, and Network Access

Apply filesystem, process, network, resource, and credential boundaries with cross-platform local fallbacks. Block path traversal, SSRF, shell injection, and unsafe downloads; audit OpenClaw sandbox and tool-policy boundaries without assuming a default install is isolated.

### 4. Protect Identity, Secrets, Tenants, and Dependencies

Use scoped short-lived credentials, delegated identity, secret stores, tenant isolation, dependency pinning, SBOMs, signatures, and vulnerability response. Treat third-party skills and plugins as supply-chain inputs with provenance, review, revocation, and a tested permission boundary.

## 4. Challenge preview: Red-Team and Repair an Exposed Agent

Exploit a deliberately vulnerable research-and-action agent, record the attack path, implement layered mitigations, and prove previously successful attacks now fail safely.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
