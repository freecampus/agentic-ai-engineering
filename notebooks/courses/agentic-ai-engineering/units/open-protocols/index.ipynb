## 1. What you will be able to do

Explain and implement pinned MCP and A2A interfaces with authorization, lifecycle management, declared conformance limits, and hostile-component defenses; distinguish protocols from skill packaging.

**Phase:** III. Architecture
**Overview:** 20–40 minutes

checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Separate Protocol, Transport, Runtime, and Framework | Layered interoperability diagram. |
| 2. Build and Inspect an MCP Server | Protocol transcript and conformance tests. |
| 3. Consume MCP with Explicit Trust | Malicious-server and authorization-boundary test suite with denied-action transcripts. |

### 1. Separate Protocol, Transport, Runtime, and Framework

Map OpenAPI, JSON Schema, MCP, A2A, HTTP, stdio, and event transport roles. Distinguish Agent Skills instruction packaging from callable tools and task protocols; diagnose diagrams that confuse discovery, execution, and authorization.

### 2. Build and Inspect an MCP Server

Expose a read-only tool and resource through a local MCP server. Inspect initialization, capability negotiation, schemas, calls, results, errors, and shutdown.

### 3. Consume MCP with Explicit Trust

Build a client that validates server identity, capabilities, schemas, outputs, and permissions. Test tool-description poisoning and capability changes; use a local HTTP authorization fixture to check token audience, consent, and rejection of token passthrough separately from stdio process trust.

### 4. Exchange Tasks Between Agents

Implement discovery, task submission, progress, artifacts, cancellation, and terminal status through an explicitly declared subset of a pinned A2A revision. Preserve caller identity and authorization context; distinguish an educational simulation from demonstrated interoperability.

## 4. Challenge preview: Build a Secure Interoperability Boundary

Connect a course agent to one local MCP server and one remote-style agent interface. Pass version mismatch, authentication failure, poisoned description, cancellation, and rate-limit tests.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
