## 1. What you will be able to do

Distinguish frameworks, agent harnesses, and complete assistant applications; migrate a stable application through adapters and quantify abstraction benefits and costs.

**Phase:** III. Architecture
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the

Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Inspect What a Framework Does for You | Framework execution map. |
| 2. Define Ports for Models, Tools, State, and Tracing | Contract test suite used by multiple adapters. |
| 3. Rebuild with a State-Graph Framework | Parity tests against the reference runtime. |
| 4. Compare a Lightweight Alternative | Framework decision record. |

### 1. Inspect What a Framework Does for You

Trace model messages, state serialization, tool dispatch, retries, callbacks, and persistence through a framework. Distinguish framework, agent harness, and complete assistant application using OpenClaw and Hermes architecture case studies; identify defaults that alter behavior.

### 2. Define Ports for Models, Tools, State, and Tracing

Refactor the plain runtime into stable domain interfaces and infrastructure adapters. Keep provider objects out of core logic.

### 3. Rebuild with a State-Graph Framework

Implement a course task with LangGraph or a current equivalent, using typed state, checkpoints, interrupts, and explicit terminal nodes.

### 4. Compare a Lightweight Alternative

Implement the same task with smolagents, PydanticAI, LlamaIndex, Haystack, or another approved open framework. Compare code volume, clarity, portability, runtime behavior, and lock-in.

## 4. Challenge preview: Migrate Without Changing the Contract

Move an agent between frameworks while preserving task interface, tool contracts, serialized state, eval dataset, and acceptance gates. Document unavoidable semantic differences.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
