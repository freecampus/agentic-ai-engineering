Explain how messages become model outputs, enforce structured responses, measure nondeterminism, and select models by evidence and license.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Follow Tokens, Messages, and Context | Rendered prompt/context report without hidden reasoning collection. |
| 2. Control Sampling and Nondeterminism | Distribution of outcomes across repeated trials. |
| 3. Make Outputs Machine-Checkable | Schema-validation suite and failure taxonomy. |
| 4. Compare Open Models Responsibly | Model-selection card containing license, hardware, quality, latency, and limitations. |

### 1. Follow Tokens, Messages, and Context

Inspect tokenization, chat templates, roles, special tokens, and context limits. Predict how message ordering and truncation change the model-visible input.

### 2. Control Sampling and Nondeterminism

Experiment with temperature, top-p, seeds, and repeated runs. Separate reproducibility of code and inputs from reproducibility of hosted model outputs.

### 3. Make Outputs Machine-Checkable

Move from prose parsing to JSON Schema, typed validation, constrained decoding, and repair or rejection policies. Test malformed, partial, and semantically invalid outputs.

### 4. Compare Open Models Responsibly

Distinguish open weights, source, data, and licenses; inspect model cards; and evaluate small models on a narrow course task. Avoid choosing from benchmark headlines alone.

## 4. Challenge preview: Build a Reliable Decision Component

Create a typed routing component backed by a fake model and at least one open model. It must reject invalid outputs, expose uncertainty through observable signals, and pass a repeated-run evaluation.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
