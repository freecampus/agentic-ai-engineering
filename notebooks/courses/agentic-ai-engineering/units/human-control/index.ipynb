## 1. What you will be able to do

Create approval, escalation, explanation, intervention, and governance mechanisms matched to uncertainty, consequence, reversibility, and operator workload.

**Phase:** IV. Evaluation and safety
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Complete the preceding unit, [Secure Agents by Construction](https://freecampus.github.io/agentic-ai-engineering/courses/agentic-ai-engineering/units/agent-security/index.html), when its lessons become available. This is a cumulative engineering sequence, not a set of independent tutorials.

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the

Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Calibrate Autonomy to Risk | Autonomy policy table tested against scenarios. |
| 2. Ask for Approval with Decision-Relevant Evidence | Approval payload and usability critique. |
| 3. Communicate Uncertainty, Provenance, and Limits | Explanation/provenance evaluation. |
| 4. Escalate, Interrupt, Appeal, and Audit | Intervention-state tests and audit record. |

### 1. Calibrate Autonomy to Risk

Classify actions by impact, reversibility, confidence, novelty, and authorization. Choose automatic, notify, confirm, dual-control, or prohibit policies.

### 2. Ask for Approval with Decision-Relevant Evidence

Present the proposed action, target, rationale, source evidence, alternatives, consequences, and rollback. Avoid approval fatigue and deceptive defaults.

### 3. Communicate Uncertainty, Provenance, and Limits

Show what the system knows, inferred, could not verify, and chose not to do. Compare confidence language with empirically measured calibration.

### 4. Escalate, Interrupt, Appeal, and Audit

Design timeouts, safe pauses, takeover, cancellation, appeals, incident records, and accountability. Handle the case where stopping immediately is itself harmful.

## 4. Challenge preview: Deliver a Defended Agent

Take the grounded knowledge worker through evaluation, observability, threat modeling, red-team testing, and human-control review. Repair the highest-risk failure and demonstrate the regression gate.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
