## 1. What you will be able to do

Compare direct response, ReAct-style interaction, planning, reflection, and search; then choose a strategy using quality, latency, and cost evidence.

**Phase:** I. Foundations
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Connect Agent Reasoning to Search and Decision Processes | Formalized small task and search tree. |
| 2. Interleave Reasoning and Acting | Trajectory grader for action relevance and observation use. |
| 3. Separate Planning from Execution | Plan schema and revision tests. |
| 4. Add Reflection and Search Only When They Help | Quality–latency–cost comparison. |

### 1. Connect Agent Reasoning to Search and Decision Processes

Use state spaces, actions, transition uncertainty, heuristics, and partial observability to explain agent behavior. Relate MDP/POMDP ideas to tool-using systems without requiring a full formal treatment.

### 2. Interleave Reasoning and Acting

Implement a ReAct-like loop using concise external action rationales rather than collecting private chain-of-thought. Detect action drift and unsupported conclusions.

### 3. Separate Planning from Execution

Build planner–executor behavior with plan validation, revision, and state updates. Test plans against unavailable tools, hidden dependencies, and changing observations.

### 4. Add Reflection and Search Only When They Help

Compare critique/revision, verifier loops, best-of-N, and tree search. Measure cases where extra inference increases confidence, latency, or correlated error without improving success.

## 4. Challenge preview: Ship a Transparent Research Assistant

Build a framework-free assistant with two read-only tools, typed state, citations, budgets, tests, traces, and a deterministic baseline. Submit an evaluation showing whether agentic control adds value.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
