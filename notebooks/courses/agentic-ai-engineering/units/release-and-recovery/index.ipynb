## 1. What you will be able to do

Version every behavior-shaping artifact, including skills and approved memory state; enforce CI gates, release gradually, monitor SLOs and drift, and respond to incidents with rollback and learning.

**Phase:** VI. Production
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Version Models, Prompts, Tools, Policies, and Data | Reproducibility check with a complete behavior-asset manifest and rollback dependency map. |
| 2. Build CI Evaluation and Security Gates | Passing pipeline and intentional-failure demonstration. |
| 3. Release with Shadows, Canaries, Flags, and Rollback | Release plan and canary analysis. |
| 4. Define SLOs, Alerts, Runbooks, and Postmortems | Incident timeline and corrective actions. |

### 1. Version Models, Prompts, Tools, Policies, and Data

Create a run manifest linking code, environment, model, harness, prompt, schemas, tool and protocol versions, policies, skills, approved memory snapshots, and dataset. Reproduce a prior evaluation and identify which behavior assets must roll back together.

### 2. Build CI Evaluation and Security Gates

Run unit, contract, replay, evaluation, adversarial, dependency, and secret checks with tiered budgets. Prevent flaky model calls from making every code change untestable.

### 3. Release with Shadows, Canaries, Flags, and Rollback

Compare versions on recorded or shadow traffic, canary a bounded population, monitor decision criteria, and roll back code and behavior assets together.

### 4. Define SLOs, Alerts, Runbooks, and Postmortems

Set indicators for task success, unsafe action, latency, cost, availability, and escalation. Respond to a simulated incident, including a poisoned skill update or unattended assistant action, and write an evidence-rich postmortem with containment and rollback.

## 4. Challenge preview: Stop a Bad Release Before It Spreads

A new version improves average quality but causes rare tool misuse and higher tail cost. Detect it through gates/canary evidence, roll back, identify the mechanism, and preserve regression tests.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
