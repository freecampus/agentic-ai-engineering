## 1. What you will be able to do

Implement a framework-free, typed agent loop with explicit state, events, budgets, termination, and deterministic tests.

**Phase:** I. Foundations
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 2. Before you begin

Complete the preceding unit, [Treat Language Models as Probabilistic Components](https://freecampus.github.io/agentic-ai-engineering/courses/agentic-ai-engineering/units/probabilistic-models/index.html), when its lessons become available. This is a cumulative engineering sequence, not a set of independent tutorials.

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the
shared package. Work in a clean Colab/Jupyter notebook or a local Conda

provider keys or paid inference.
Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Represent State and Events Explicitly | Serialized trajectory that can be inspected and replayed. |
| 2. Implement the Decide–Act–Observe Loop | Tested runtime with fake model and tool ports. |
| 3. Make Termination and Budgets Enforceable | Tests proving every budget terminates execution. |
| 4. Stream, Cancel, and Handle Async Work | Cancellation and timeout integration tests. |

### 1. Represent State and Events Explicitly

Design typed task, message, tool-call, observation, budget, and result records. Compare mutable hidden state with an append-only event history and derived state.

### 2. Implement the Decide–Act–Observe Loop

Write the smallest complete agent loop in plain Python. Identify the model, agent harness, tools, and environment separately; trace normal execution, a tool error, and an unavailable capability.

### 3. Make Termination and Budgets Enforceable

Add maximum steps, time, tokens, cost, repeated-action detection, and explicit success/failure states. Diagnose loops that merely ask the model to stop itself.

### 4. Stream, Cancel, and Handle Async Work

Introduce async model/tool calls, progress events, cancellation, deadlines, and cleanup. Repair a runtime that leaks a task after cancellation.

## 4. Challenge preview: Complete the Transparent Runtime

Extend a partial runtime to survive malformed decisions, tool timeouts, repeated actions, cancellation, and budget exhaustion while preserving a complete event log.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
