## 1. What you will be able to do

Assemble a bounded, provenance-aware context from instructions, state, data, tools, and skill packages while managing discovery, conflicts, trust, and compression.

**Phase:** II. Tools and knowledge
**Overview:** 20–40 minutes
**Planned unit effort:** 8–14 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Classify Context by Purpose and Trust | Context manifest with source and trust labels. |
| 2. Select and Order the Working Set | Context-selection evaluation with discovery recall, token accounting, and task-success tradeoffs. |
| 3. Compress Without Inventing State | Compression fidelity tests. |
| 4. Detect Conflict, Contamination, and Context Rot | Adversarial context suite. |

### 1. Classify Context by Purpose and Trust

Separate system policy, developer instruction, user request, retrieved data, tool output, memory, and generated artifacts. Classify repository instruction files and SKILL.md procedures by provenance; neither a filename nor skill metadata grants execution authority.

### 2. Select and Order the Working Set

Implement relevance, recency, authority, diversity, and token-budget rules. Compare eager loading with tool discovery and progressive disclosure of skill descriptions, instructions, and references; measure ordering sensitivity and missed capabilities as well as token savings.

### 3. Compress Without Inventing State

Compare truncation, extractive selection, structured summaries, and hierarchical compression. Preserve decisions, unresolved questions, provenance, and constraints across context resets; distinguish a readable handoff from authoritative runtime state.

### 4. Detect Conflict, Contamination, and Context Rot

Introduce conflicting instructions, stale records, and malicious retrieved text. Build policies for exclusion, quarantine, user clarification, and source-aware disagreement.

## 4. Challenge preview: Rescue an Overloaded Context Window

Reduce an oversized, contradictory context while maintaining task success and citations. Demonstrate lower token usage and equal or better held-out performance.

The complete challenge contract, starting fixtures, progressive checks, three-level
hint ladder, and solution path are not authored yet. Do not interpret this outline
as an executable assessment. A successful-looking output alone will not establish
reliability: preserve the inputs, limits, observed behavior, and failed cases too.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.
