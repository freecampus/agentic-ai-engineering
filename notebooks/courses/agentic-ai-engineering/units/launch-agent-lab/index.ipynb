# Launch a Reproducible Agent Engineering Lab Overview

Generated from FreeCampus Agentic AI Engineering. Run cells from top to bottom, write predictions before execution, and change one thing at a time.

[Source page](https://freecampus.github.io/agentic-ai-engineering/courses/agentic-ai-engineering/units/launch-agent-lab/index.html)

## Curriculum preview—not a released unit
The 5 lessons and challenge have complete authoring drafts awaiting independent
teaching and accessibility review; they remain planned for publication. This
overview does not award course progress. Its notebook is an orientation copy,
not a completed instructional lab.

## 1. What you will be able to do

Run, inspect, and modify a small deterministic agent in Colab; reproduce it locally; keep secrets out of artifacts; and distinguish model, runtime, tool, environment, and trace evidence.

**Phase:** Orientation
**Overview:** 20–40 minutes
**Planned unit effort:** 10–16 hours, including typing, experiments,
checks, and a 90–180 minute challenge. Estimates will be recalibrated during authoring.

**Scenario:** help a repair café hand off a dependable launch kit. A correct-looking
sum, a missing document, an unsafe suggested fix, a stale notebook, and a broken
local entry point each expose a different kind of evidence failure. Your final
artifact is a launch card another volunteer can reproduce—not a screenshot of 12.

## 2. Before you begin

Bring project-ready Python: functions, data structures, modules, exceptions, and basic tests. Use the [readiness guide](https://freecampus.github.io/agentic-ai-engineering/courses/agentic-ai-engineering/readiness.html) to check gaps.

Use the [offline smoke lab](https://freecampus.github.io/agentic-ai-engineering/resources/agent-lab.html) to check the
shared package. Work in a clean Colab/Jupyter notebook or a local Conda
environment with Poetry-managed dependencies. The core route will not require
provider keys or paid inference.
Keep credentials and personal data out of notebooks, traces, and submitted artifacts.

The notebook route bundles a small offline package and uses only synthetic data.
Local development requires Git and the real project Conda + Poetry environment;
initial installation needs registry access. No GPU is required. Use an explicit
workspace, keep your journal outside temporary runtime storage, and distinguish
notebook success from a verified local installation.

**Working vocabulary:** the task states the requested result; the model proposes
an action; the runtime authorizes and dispatches it; the tool returns data; the
trajectory records visible events. The model here is a deterministic test double,
not an LLM. The in-process examples are not security sandboxes. General runtime,
tool-boundary, and security engineering arrive in Units 3, 5, and 15.

## 3. Planned lesson sequence

| Lesson | Evidence you will preserve |
|---|---|
| 1. Meet an Agent Through a Tiny Complete Run | Annotated event log connecting task, model decision, tool call, observation, and final answer. |
| 2. Learn with Specifications and Evidence | Task contract plus a prediction/run/explanation record. |
| 3. Use AI Assistance Without Outsourcing Judgment | Verification checklist and corrected implementation rationale. |
| 4. Work Reliably in Colab and Jupyter | Clean restart-and-run transcript. |
| 5. Build a Dependable Local Workspace | Environment report and passing cross-environment smoke test. |

### 1. Meet an Agent Through a Tiny Complete Run

Run a three-step deterministic agent that reads a task, chooses a calculator tool, and returns a result. Learners map every source object to the visible trajectory, predict one controlled edit, and repair a malformed tool result.

### 2. Learn with Specifications and Evidence

Turn a vague request into a task contract containing success, stop, budget, escalation, and forbidden-action rules. Compare a convincing demo with a reproducible test and learn why one successful run is not evidence of reliability.

### 3. Use AI Assistance Without Outsourcing Judgment

Establish a responsible assistance protocol for course work: disclose assistance, protect data and credentials, verify generated claims and code, and preserve the learner’s own reasoning. Diagnose a confident but unsafe suggested tool implementation.

### 4. Work Reliably in Colab and Jupyter

Inspect notebook state, execution order, package installation, files, runtime resets, and reproducibility. Recover a notebook whose apparent success depends on stale variables and an unrecorded package version.

### 5. Build a Dependable Local Workspace

Create the course environment with Git, Conda, `Poetry`, Python, pytest, Ruff, type checking, and environment-variable configuration on Windows, macOS, or Linux. Run the same smoke test locally and in Colab.

## 4. Challenge preview: Prove the Lab Is Ready

Repair a deliberately stale notebook and broken local project, run the tiny agent from clean state, reject an unsafe AI-generated fix, and submit a launch card containing versions, commands, outputs, and unresolved limitations.

The draft challenge includes a deliberately stale notebook, a CLI that ignores its
input path, independent fresh-process checks, three progressive hints, and a
hidden solution. It is not yet a released assessment. The contract requires
explicit input paths, current operands, denial before tool effects, and an
actionable failure instead of a fabricated missing input. Local setup and
clean-kernel evidence are separate gates. A successful-looking output alone will
not establish reliability: preserve inputs, limits, behavior, and failed cases.

## 5. Suggested stopping points

Plan one session per lesson. Before changing an implementation, write a prediction;
after each session, record what ran, what failed, what changed, and what remains
unverified. Save a clean run and an architecture note before attempting the challenge.
Finish with a retrospective on quality, failures, latency, cost, and safety.

Suggested breaks: after the first annotated trace (Lesson 1); after the negative
lookup test and assistance disclosure (Lessons 2–3); after a clean notebook run
(Lesson 4); and after the local environment report (Lesson 5). Save evidence at
each break. Pause setup work on an unavailable registry rather than silently
substituting a different environment. The challenge finishes with a comparison
against direct arithmetic and an explicit list of unverified claims.
