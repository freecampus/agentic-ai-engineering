## 1. Separate returning a result from displaying it

Before running, predict which call returns data to its caller.

```python
def display_total(left, right):
    print(left + right)


def add(left, right):
    return left + right


shown = display_total(2, 3)
returned = add(2, 3)
assert shown is None
assert returned == 5
```

The first function displays `5`, then returns `None` implicitly. The second returns
an integer without displaying it. A tool consumer needs data it can inspect, not
just a message printed to a terminal. Change the two inputs and update the

## 2. Track shared mutable state

Predict the final list before running. Does assigning a second name copy the list?

```python
events = ["task"]
shared = events
shared.append("tool_call")
assert events == ["task", "tool_call"]
```

Both names refer to the same list. Replace `shared = events` with
`shared = events.copy()`. Now `events` should remain `["task"]`; change the assertion
to reflect the intended contract. Explain why hidden sharing could corrupt a trace.

## 3. Locate the failure before changing the expected value

Predict the exception produced by the conversion, then inspect its exact message.

```python
try:
    value = int("twelve")
except ValueError as error:
    print(type(error).__name__, str(error))
```

Python raises `ValueError` because this text does not represent an integer.
Handling it does not create a valid answer. A real caller needs a policy such as
rejecting input, asking for clarification, or reporting failure.
